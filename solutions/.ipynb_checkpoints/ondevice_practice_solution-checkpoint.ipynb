## 📌 개념 1. Magnitude-based Pruning (Fine-grained)

**핵심 아이디어**: "절댓값이 작은 weight는 출력에 기여가 작다"고 가정하고, 그 중 일정 비율을 0으로 만듭니다.

**Granularity(세밀함) 계층**:
- **Fine-grained (Element-wise)**: 개별 weight 하나씩 제거 — 가장 높은 유연성, 높은 정확도
- **Vector-level**: 한 행(row) 단위 제거
- **Kernel-level**: conv kernel (`kh × kw`) 단위 제거
- **Channel-level**: 채널 통째로 제거 — 하드웨어 가속 friendly

**공식**:
1. 제거할 개수 = `round(total_elements × sparsity)`
2. importance = `|W|` (절댓값)
3. threshold = importance 중 하위에서 `num_pruned`번째 값 (`torch.kthvalue`)
4. mask = `importance > threshold`

`torch.kthvalue(x, k)`는 x에서 **k번째로 작은 값**을 반환합니다.

### 문제 1. Fine-grained Magnitude Pruning

```python
def prune_weight_fine_grained(weight: torch.Tensor, sparsity: float) -> torch.Tensor:
    sparsity = min(1.0, max(0.0, sparsity))
    if sparsity == 1.0:
        weight.zero_()
        return torch.zeros_like(weight)
    elif sparsity == 0.0:
        return torch.ones_like(weight)

    # 전체 원소 중 sparsity 비율만큼 제거할 개수를 계산
    num_pruned_elements = ### round(weight.numel() * sparsity) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    # 각 weight의 중요도 = 절댓값 (magnitude 기준)
    importance = ### torch.abs(weight) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    # 하위 num_pruned_elements번째로 작은 값이 cutoff threshold
    # flatten() 후 kthvalue를 적용해 전체 분포에서 기준값을 구함
    threshold = ### torch.kthvalue(importance.flatten(), num_pruned_elements)[0] ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    # threshold 초과는 살리고(1), 이하는 제거(0)
    mask = ### importance > threshold ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    weight.mul_(mask)
    return mask
```

## 📌 개념 2. Kernel-level Pruning

Conv weight shape: `(out_channels, in_channels, kh, kw)`. Kernel 단위로 제거하므로 **커널 1개 = `(kh × kw)` matrix 하나**를 통째로 0으로 만듭니다.

**차이점**:
- Fine-grained: `numel()` 전체를 기준으로 세어 element별 제거
- Kernel-level: 커널 개수 = `out_channels × in_channels`, 중요도는 `(kh, kw)` 차원에 대해 합산
- `keepdim=True`로 남겨서 이후 `expand_as(weight)`로 shape을 맞추는 것이 관건

### 문제 2. Kernel-level Pruning

```python
def prune_weight_kernel_level(weight: torch.Tensor, sparsity: float) -> torch.Tensor:
    sparsity = min(1.0, max(0.0, sparsity))
    if sparsity == 1.0:
        weight.zero_()
        return torch.zeros_like(weight)
    elif sparsity == 0.0:
        return torch.ones_like(weight)

    # 전체 커널 개수 = out_channels × in_channels (커널 하나 = kh×kw 매트릭스)
    num_kernels = ### weight.shape[0] * weight.shape[1] ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    num_pruned_kernels = ### round(num_kernels * sparsity) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    # 각 커널 내부 (kh, kw)에 대해 |W|의 합을 구해 중요도 계산
    # keepdim=True로 남겨야 이후 expand_as()로 원본 shape 복원 가능
    importance = ### weight.abs().sum(dim=(2, 3), keepdim=True) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    threshold = torch.kthvalue(importance.flatten(), num_pruned_kernels)[0]
    mask = importance > threshold
    mask = mask.expand_as(weight)
    weight.mul_(mask)
    return mask
```

## 📌 개념 3. Global vs Layer-wise Pruning

- **Layer-wise**: 모든 레이어에 동일(또는 레이어별 지정) sparsity 적용
- **Global**: 전 레이어 weight를 모두 concat하여 **하나의 threshold**로 판정

Global이 더 유연하지만 (중요한 레이어는 적게 prune, 덜 중요한 레이어는 많이 prune), 구현에는 모든 weight를 1D로 붙이는 `torch.cat` 단계가 필요합니다.

### 문제 3. Global Magnitude Pruning

```python
def global_prune(model, sparsity):
    parameters_to_prune = []
    for name, param in model.named_parameters():
        if param.dim() > 1:
            parameters_to_prune.append(param.view(-1))

    # 모든 레이어의 weight를 하나의 1D 텐서로 concat → 전역 비교 가능
    all_weights = ### torch.cat(parameters_to_prune) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    num_elements = all_weights.numel()
    num_zeros = ### round(num_elements * sparsity) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    importance = ### torch.abs(all_weights) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    # 전역 threshold: 모든 레이어를 통틀어 하위 num_zeros번째 값
    threshold = ### torch.kthvalue(importance, num_zeros)[0] ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    masks = {}
    for name, param in model.named_parameters():
        if param.dim() > 1:
            masks[name] = torch.abs(param.data) > threshold
    return masks
```

## 📌 개념 4. Sparsity Schedule (Iterative Pruning)

한 번에 높은 sparsity를 적용하면 모델이 망가지므로, 에폭마다 점진적으로 늘리는 **스케줄링**을 사용합니다.

**공식** (원 논문 [Zhu & Gupta 2017]):
$$ s(t) = s_f + (s_i - s_f) \left(1 - \frac{t - t_0}{t_e - t_0}\right)^n $$

- `exponent n=1`: Linear schedule
- `exponent n=3`: Cubic schedule (초반에 빨리 늘고 말미에 평탄)

구간별 처리:
- `epoch < epoch_start`: 아직 pruning 시작 전 → 0
- `epoch >= epoch_end`: 종료된 이후 → `sparsity_end` 유지
- 그 사이: 위 공식 적용

### 문제 4. Sparsity Schedule

```python
def get_sparsity_schedule(num_epochs, epoch_start, epoch_end,
                          sparsity_start, sparsity_end, exponent):
    sparsity_schedule = []
    for epoch in range(num_epochs):
        if epoch < epoch_start:
            # 아직 pruning이 시작되지 않음
            sparsity = ### 0 ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
        elif epoch >= epoch_end:
            # 목표 sparsity에 도달한 뒤 유지
            sparsity = ### sparsity_end ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
        else:
            # s_f + (s_i - s_f) * (1 - (t-t0)/(te-t0))^n
            # exponent=1이면 linear, =3이면 cubic
            sparsity = ### sparsity_end + (sparsity_start - sparsity_end) * (1 - (epoch - epoch_start) / (epoch_end - epoch_start)) ** exponent ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
        sparsity_schedule.append(sparsity)
    return sparsity_schedule
```

## 📌 개념 5. Linear Quantization 공식

FP → INT 변환 기본 공식:
$$ q = \text{round}(x / s) + z $$
$$ x \approx (q - z) \cdot s $$

- `s`: scale, `z`: zero_point
- 4단계: (1) scale로 나눔 → (2) round → (3) zero_point shift → (4) bitwidth 범위로 clamp

### 문제 5. Linear Quantization

```python
def linear_quantize(fp_tensor, bitwidth, scale, zero_point, dtype=torch.int8):
    # Step 1: scale로 나눠 정수 범위로 스케일
    scaled_tensor = ### fp_tensor / scale ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    # Step 2: 정수로 반올림
    rounded_tensor = ### torch.round(scaled_tensor) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    rounded_tensor = rounded_tensor.to(dtype)
    # Step 3: zero_point 만큼 shift (비대칭 양자화의 경우 fp=0에 대응시키는 역할)
    shifted_tensor = ### rounded_tensor + zero_point ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    # Step 4: 대상 bitwidth 범위로 자름
    quantized_min, quantized_max = get_quantized_range(bitwidth)
    quantized_tensor = shifted_tensor.clamp_(quantized_min, quantized_max)
    return quantized_tensor
```

## 📌 개념 6. Scale & Zero Point

FP `[fp_min, fp_max]`를 INT `[q_min, q_max]`로 매핑할 때:

$$ s = \frac{\text{fp\_max} - \text{fp\_min}}{q_{\max} - q_{\min}} $$

fp=0을 정수축의 어디로 보낼지 결정하는 값이 `zero_point`:
$$ z = \text{round}(q_{\min} - \text{fp\_min} / s) $$

주의: `q_max - q_min = 2^bitwidth - 1`

### 문제 6. Scale & Zero Point 계산

```python
def get_quantization_scale_and_zero_point(fp_tensor, bitwidth):
    quantized_min, quantized_max = get_quantized_range(bitwidth)
    fp_max = fp_tensor.max().item()
    fp_min = fp_tensor.min().item()

    # FP 범위를 INT 범위로 선형 매핑하는 scale
    scale = ### (fp_max - fp_min) / (quantized_max - quantized_min) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    # fp=0이 정수축에서 어디인지 (비대칭 양자화)
    zero_point = ### round(quantized_min - fp_min / scale) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    if zero_point < quantized_min:
        zero_point = quantized_min
    elif zero_point > quantized_max:
        zero_point = quantized_max
    else:
        zero_point = round(zero_point)
    return scale, int(zero_point)
```

## 📌 개념 7. Quantized Linear/Conv Output 재스케일

양자화된 `input × weight` (정수 연산)의 결과는 output 양자화 범위에 맞게 **다시 스케일**해야 합니다.

$$ y_q = \text{round}\left( y_{\text{int}} \cdot \frac{s_{\text{input}} \cdot s_{\text{weight}}}{s_{\text{output}}} \right) + z_{\text{output}} $$

- per-channel weight의 경우 `weight_scale`이 벡터이므로 broadcasting을 위해 shape 조정 필요

### 문제 7. Quantized Linear Output 보정

```python
def quantized_linear_core(output, input_scale, weight_scale, output_scale, output_zero_point):
    weight_scale = weight_scale.flatten().view(1, -1)
    output = output.float()

    # Step 2: (s_input * s_weight / s_output)으로 재스케일 → output 양자화 grid에 align
    output = ### output * (input_scale * weight_scale / output_scale) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    # Step 3: output zero_point 만큼 shift
    output = ### output + output_zero_point ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    return output
```

## 📌 개념 8. K-means Quantization

선형 양자화와 달리 **비균등** 양자화. FP weight 값들을 K-means로 `2^bitwidth`개의 클러스터로 묶고, 각 weight를 자기 클러스터의 centroid 값으로 치환합니다.

- codebook = (centroids, labels)
- `centroids[labels]`로 원래 shape으로 decode

### 문제 8. K-means Quantization

```python
def k_means_quantize(fp32_tensor, bitwidth=4, codebook=None):
    if codebook is None:
        # bitwidth 비트로 표현 가능한 클러스터 수 = 2^bitwidth
        n_clusters = ### 2 ** bitwidth ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
        kmeans = KMeans(n_clusters=n_clusters, mode='euclidean', verbose=0)
        labels = kmeans.fit_predict(fp32_tensor.view(-1, 1)).to(torch.long)
        centroids = kmeans.centroids.to(torch.float).view(-1)
        codebook = Codebook(centroids, labels)

    # 디코딩: labels가 가리키는 centroid 값으로 weight 대체
    quantized_tensor = ### codebook.centroids[codebook.labels] ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    fp32_tensor.set_(quantized_tensor.view_as(fp32_tensor))
    return codebook
```

## 📌 개념 9. Centroid 업데이트 (K-means fine-tuning)

Fine-tune 중 weight가 바뀌면 centroid도 갱신해야 품질이 유지됩니다. 각 클러스터 k에 속한 weight의 **평균**이 새 centroid.

### 문제 9. K-means Codebook 업데이트

```python
def update_codebook(fp32_tensor, codebook):
    n_clusters = codebook.centroids.numel()
    fp32_tensor = fp32_tensor.view(-1)
    for k in range(n_clusters):
        # label==k인 원소들의 평균값이 새 centroid
        codebook.centroids[k] = ### torch.mean(fp32_tensor[codebook.labels == k]) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
```

## 📌 개념 10. Knowledge Distillation (Soft Targets)

작은 Student가 큰 Teacher의 **soft 확률 분포**를 모방하게 학습. Hard label만 쓰는 CE보다 "클래스 간 유사도" 정보를 더 얻습니다.

**Temperature T**: softmax를 부드럽게. T가 클수록 분포가 평탄해져 dark knowledge 전달.

**Loss 공식** (Hinton 2015):
$$ \mathcal{L}_{KD} = T^2 \cdot \text{KL}(P_T \| P_S) \quad , \quad P = \text{softmax}(\text{logits}/T) $$

$T^2$ 곱하는 이유: softmax 미분에서 생기는 $1/T^2$ 항을 보정하기 위함.

### 문제 10. KD Soft Targets Loss

```python
def kd_loss_step(teacher, student, inputs, labels, T,
                 soft_target_loss_weight, ce_loss_weight, ce_loss):
    with torch.no_grad():
        teacher_logits = teacher(inputs)
    student_logits = student(inputs)

    # Temperature T로 softmax를 부드럽게 만들기
    soft_targets = ### nn.functional.softmax(teacher_logits / T, dim=-1) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    student_prob = ### nn.functional.softmax(student_logits / T, dim=-1) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    # KL(P_T || P_S) = Σ P_T (log P_T − log P_S) ; 배치 평균 후 T^2 스케일링
    soft_targets_loss = ### torch.sum(soft_targets * (soft_targets.log() - student_prob.log())) / student_prob.size(0) * (T ** 2) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    label_loss = ce_loss(student_logits, labels)
    loss = soft_target_loss_weight * soft_targets_loss + ce_loss_weight * label_loss
    return loss
```

## 📌 개념 11. Cosine Similarity KD (Representation-level)

Logits가 아닌 **hidden feature 벡터**를 teacher-student 간 정렬. `CosineEmbeddingLoss(target=1)`이면 loss 최소화 = cosine similarity 최대화.

### 문제 11. Cosine Loss KD

```python
def cosine_kd_step(teacher, student, inputs, labels,
                   hidden_rep_loss_weight, ce_loss_weight, ce_loss, cosine_loss):
    with torch.no_grad():
        _, teacher_hidden = teacher(inputs)
    student_logits, student_hidden = student(inputs)

    # target=1인 CosineEmbeddingLoss는 두 벡터의 cosine sim을 1에 가깝게 만듬
    hidden_rep_loss = ### cosine_loss(student_hidden, teacher_hidden, target=torch.ones(inputs.size(0)).cuda()) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    label_loss = ce_loss(student_logits, labels)
    loss = hidden_rep_loss_weight * hidden_rep_loss + ce_loss_weight * label_loss
    return loss
```

## 📌 개념 12. Hint-based KD (MSE on Feature Map)

Teacher와 Student의 중간 feature map을 **MSE**로 정렬. Student의 채널 수가 적으면 trainable regressor(예: 1x1 conv)로 차원 맞춤.

### 문제 12. MSE Feature Map KD

```python
def mse_kd_step(teacher, student, inputs, labels,
                feature_map_weight, ce_loss_weight, ce_loss, mse_loss):
    with torch.no_grad():
        _, teacher_feature_map = teacher(inputs)
    student_logits, regressor_feature_map = student(inputs)

    # Student regressor 출력과 teacher feature map 간 MSE
    hidden_rep_loss = ### mse_loss(regressor_feature_map, teacher_feature_map) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    label_loss = ce_loss(student_logits, labels)
    loss = feature_map_weight * hidden_rep_loss + ce_loss_weight * label_loss
    return loss
```

## 📌 개념 14. Wanda Activation Norm 수집

Wanda의 중요도 = `|W| × ||X||_2`. 즉 weight 크기만이 아니라, 해당 채널로 들어오는 **입력 activation의 크기**까지 고려합니다.

Calibration 데이터를 여러 batch에 나눠 누적할 때 L2 norm을 그대로 누적하면 틀리므로, **제곱을 누적한 뒤 마지막에 sqrt**합니다:
- 누적: `||x||^2 = Σ x_i^2` (dim=1은 batch/token 축)
- 최종: `sqrt(누적값)`
- x.shape = `(hidden, batch)` → `torch.norm(x, p=2, dim=1)` 으로 hidden별 L2 계산

### 문제 14. Wanda Activation Norm

```python
def compute_activation_norm(x):
    x = x.type(torch.float32)
    # L2 norm^2 형태로 누적 가능 (이후 sqrt_()로 최종 L2)
    # dim=1은 batch/token 축, 결과 shape=(hidden_size,)
    activation_norm = ### torch.norm(x, p=2, dim=1) ** 2 ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    return activation_norm
```

## 📌 개념 15. Wanda Pruning

**핵심 차이**: magnitude-based는 행렬 전체에서 threshold 하나. Wanda는 **row-wise** 동일 sparsity.

- importance $S_{ij} = |W_{ij}| \cdot \|X_j\|_2$
- 각 row 별로 하위 `col × sparsity`개를 제거
- `torch.kthvalue(..., dim=1)` → row별 k번째 값 반환

### 문제 15. Wanda Pruning

```python
@torch.no_grad()
def prune_wanda(model, sparsity, input_feat):
    for n, m in model.named_modules():
        if isinstance(m, nn.Linear) and "lm_head" not in n:
            W = m.weight.data
            row, col = W.shape
            # row당 제거할 원소 개수
            num_zeros_per_row = ### round(col * sparsity) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
            # 중요도 = |W| * activation_norm  (Wanda 핵심)
            importance = ### torch.abs(W) * input_feat[n] ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
            # 각 row 별로 독립 threshold → (row,) 형태 반환
            threshold = ### torch.kthvalue(importance, num_zeros_per_row, dim=1)[0] ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
            # broadcasting 위해 (row, 1)로 변형하여 row-wise 비교
            mask = ### importance > threshold.reshape(row, 1) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
            W.mul_(mask)
```

## 📌 개념 16. AWQ: Outlier Channel 보호

Activation 중 **상위 1% 큰 채널**은 양자화 시 정보 손실에 민감. AWQ는 이런 outlier 채널의 weight를 `scale_factor`로 **키운 뒤 양자화 → 복원**해 상대적으로 높은 해상도로 표현합니다.

- `torch.topk(importance, k)` → `(values, indices)`. `[1]`로 indices 선택.

### 문제 16. AWQ outlier mask

```python
@torch.no_grad()
def pseudo_quantize_model_weight_scaleup(model, w_bit, q_group_size, input_feat, scale_factor):
    for n, m in model.named_modules():
        if isinstance(m, nn.Linear):
            importance = sum(input_feat[n]).float()
            # 상위 1% activation 채널의 인덱스 선택
            outlier_mask = ### torch.topk(importance, int(len(importance) * 0.01))[1] ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
            assert outlier_mask.dim() == 1

            # 중요한 채널은 scale_factor로 확대 후 양자화 → 작은 값은 덜 깎임
            m.weight.data[:, outlier_mask] *= scale_factor
            m.weight.data = pseudo_quantize_tensor(m.weight.data, n_bit=w_bit, q_group_size=q_group_size)
            # 확대분 원상복구
            m.weight.data[:, outlier_mask] ### /= scale_factor ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
```

## 📌 개념 17. AWQ Auto-Scale (Grid Search)

고정 scale_factor 대신, **활성값 통계** $s_x$를 바탕으로 `scales = s_x ^ ratio` 공식을 사용. 여러 ratio를 grid-search해 원본 출력과 가장 가까운 결과를 내는 scale 선택.

양자화 전 `*= scales`, 양자화 후 `/= scales`로 복원.

### 문제 17. AWQ Grid Search Scale

```python
def awq_scale_step(s_x, ratio, fc, w_bit, q_group_size):
    # s_x^ratio : activation 통계에 기반한 scale
    scales = ### s_x ** ratio ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    scales = scales / (scales.max() * scales.min()).sqrt().view(1, -1)

    scales = scales.to(fc.weight.device)
    # 양자화 전에는 weight를 scale로 확대
    fc.weight.data ### *= scales ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    fc.weight.data = pseudo_quantize_tensor(fc.weight.data, w_bit, q_group_size)
    # 양자화 후 다시 축소해 원래 scale 복원
    fc.weight.data ### /= scales ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
```

## 📌 개념 18. SmoothQuant: LN-FC Smoothing

수학적 동치 변환: $Y = (X \oslash s) (s \odot W)$. LN 출력에 `/scale`을 선적용(weight, bias에 흡수)하고, 뒤따르는 FC의 weight에는 `×scale`을 곱함 → forward 결과는 동일하되 activation은 smoothing.

### 문제 18. SmoothQuant LN-FC Smoothing

```python
@torch.no_grad()
def smooth_ln_fcs_by_scale(ln, fcs, scale):
    if not isinstance(fcs, list):
        fcs = [fcs]

    # LN의 weight / bias에 scale 역적용 (출력 activation이 scale만큼 축소됨)
    ln.weight ### .div_(scale) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    ln.bias ### .div_(scale) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

    for fc in fcs:
        # FC weight 쪽에서 같은 scale을 곱해 상쇄 → 수학적으로 forward 동등
        fc.weight ### .mul_(scale) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
```

## 📌 개념 19. SmoothQuant Alpha 공식

Activation과 Weight 각각의 크기에 따라 난이도를 α로 분배:

$$ s_j = \frac{\max(|X_j|)^{\alpha}}{\max(|W_j|)^{1-\alpha}} $$

- α=1: activation 양자화 어려움 전부 weight로 이동
- α=0: weight 양자화 어려움 전부 activation으로 이동
- α=0.5: 보통 균형점

### 문제 19. SmoothQuant Alpha Scale

```python
def smooth_scale(act_scales, weight_scales, alpha=0.5):
    # s = act^α / w^(1-α)  ;  α로 activation/weight 간 양자화 난이도 균형
    scales = ### act_scales.pow(alpha) / weight_scales.pow(1 - alpha) ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
    return scales
```

## 📌 개념 20. QuaRot / SpinQuant: Weight Rotation

Weight를 직교 행렬 R로 회전시켜 **outlier를 퍼뜨려** 양자화 친화 분포로 변환. 직교 행렬이므로 forward 결과를 보존 가능 (층간 상쇄).

PyTorch의 `nn.Linear.weight`는 $(out, in)$로 저장되어 `y = x @ W.T`. 회전 시:
- **Embedding** $(vocab, hidden)$: `W @ R` → 입력 hidden을 회전
- **입력 회전 대상 Linear** (Q, K, V, gate, up): `W @ R` (weight의 in 축에 R 적용, 즉 `x @ R.T` 효과)
- **출력 회전 대상 Linear** (o_proj, down_proj): `R.T @ W` (out 축에 R.T 적용 → 다음 layer 입력의 R과 상쇄)

### 문제 20. QuaRot Rotation

```python
def rotate_model_weight(model, R1):
    for n, m in model.named_modules():
        if isinstance(m, nn.Embedding):
            # Embedding: (vocab, hidden) × (hidden, hidden)
            W_ = m.weight.data
            m.weight.data = ### W_ @ R1 ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!

        if isinstance(m, nn.Linear):
            if "o_proj" in n or "down_proj" in n:
                # 출력측: R.T를 out 축에 적용해 다음 layer의 R과 상쇄
                W_ = m.weight.data
                m.weight.data = ### R1.T @ W_ ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
            else:
                # 입력측 (Q, K, V, gate, up): in 축 회전
                W_ = m.weight.data
                m.weight.data = ### W_ @ R1 ### !!!!!!!!!!!!코드를 작성하세요!!!!!!!!!!!!
        torch.cuda.empty_cache()
```
