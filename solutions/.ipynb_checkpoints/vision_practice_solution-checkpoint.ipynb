각 문제마다 **핵심 개념 설명 → 코드 정답 + 풍부한 주석** 순으로 구성되어 있습니다.

## 문제 1. CIFAR-10 Normalize (정규화)

### 핵심 개념: 이미지 정규화(Normalization)

- **정의(definition):** 이미지의 각 채널(R,G,B) 픽셀 값을 해당 채널의 평균(mean)과 표준편차(std)를 이용해 **평균 0·표준편차 1**에 가깝게 변환하는 전처리 과정입니다. 수식: `x_new = (x - mean) / std`
- **원리:** 딥러닝 모델은 입력 분포가 일정할 때 학습이 안정되고 수렴이 빠릅니다. 정규화는 채널 간 scale 차이를 줄여 가중치 업데이트가 한쪽으로 쏠리는 것을 방지하고, 기울기 폭발/소실을 완화합니다.
- **예시:** CIFAR-10의 통계치는 `mean=(0.4914, 0.4822, 0.4465)`, `std=(0.2023, 0.1994, 0.2010)`입니다. 정규화 후 값의 범위는 대략 `-2.5 ~ 2.5`가 됩니다. ImageNet으로 사전학습된 모델을 쓴다면 보통 ImageNet 통계치 `(0.485,0.456,0.406)/(0.229,0.224,0.225)`를 사용합니다.


```python
import torchvision.transforms as T

# CIFAR-10 데이터셋에서 널리 쓰이는 채널별 평균/표준편차
CIFAR10_MEAN = (0.4914, 0.4822, 0.4465)  # R,G,B 평균
CIFAR10_STD  = (0.2023, 0.1994, 0.2010)  # R,G,B 표준편차

# 학습 데이터 전처리 파이프라인
train_tfms = T.Compose([
    T.RandomCrop(32, padding=4),          # 4픽셀 패딩 후 32x32로 랜덤 크롭(증강)
    T.RandomHorizontalFlip(),             # 좌우반전 증강
    T.ToTensor(),                         # PIL -> Tensor, [0,1] 스케일
    T.Normalize(CIFAR10_MEAN, CIFAR10_STD)  # (x - mean) / std 로 정규화
])
```

## 문제 2. Top-1 정확도(Accuracy) 계산

### 핵심 개념: Top-1 정확도와 argmax

- **정의(definition):** 모델이 출력한 로짓(logits) 중 가장 큰 값을 갖는 클래스 하나를 예측값으로 삼고, 정답(label)과 일치한 비율을 구한 지표가 **Top-1 Accuracy**입니다.
- **원리:** 로짓 텐서의 shape는 `(B, C)` (B=배치 크기, C=클래스 수)입니다. `argmax(dim=1)`을 하면 클래스 축에서 최댓값의 인덱스를 반환해 `(B,)` shape의 예측 클래스가 됩니다. 이를 정답과 비교해 평균을 내면 정확도가 나옵니다.
- **예시:** logits = `[[0.1, 2.3, 0.5], [1.0, 0.2, 3.1]]` → argmax(dim=1) = `[1, 2]`. 정답이 `[1, 0]`이라면 정확도는 1/2 = 0.5.


```python
def accuracy_top1(logits, targets):
    # logits shape: (B, num_classes) -> 클래스 차원(dim=1)에서 최댓값 인덱스 추출
    preds = logits.argmax(dim=1)  # (B,) shape 의 예측 클래스
    # 정답과 같으면 1, 다르면 0 -> float 로 변환 후 평균 = 맞춘 비율
    return (preds == targets).float().mean().item()  # Python scalar 반환
```

## 문제 3. 학습 루프의 핵심 3단계 (optimizer.zero_grad / backward / step)

### 핵심 개념: PyTorch 학습 루프의 3단계

- **정의(definition):** PyTorch에서 한 번의 파라미터 업데이트는 `optimizer.zero_grad() → loss.backward() → optimizer.step()`의 3단계로 구성됩니다.
- **원리:**
    1. `zero_grad()`: 이전 배치에서 계산돼 쌓여있는 gradient를 0으로 초기화합니다. PyTorch는 기본적으로 gradient를 **누적**하기 때문에 매 배치마다 초기화하지 않으면 이전 배치의 gradient가 더해져 잘못된 방향으로 학습됩니다.
    2. `backward()`: 현재 loss에 대해 역전파(backpropagation)를 수행해 각 파라미터의 gradient(∂loss/∂w)를 계산합니다.
    3. `step()`: 계산된 gradient를 이용해 optimizer의 업데이트 규칙(SGD, Adam 등)대로 파라미터를 갱신합니다. 예) SGD: `w = w - lr * ∂loss/∂w`.
- **예시:** 이 순서를 지키지 않고 `backward()`만 반복 호출하면 gradient가 무한히 쌓이는 버그가 발생합니다.


```python
def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()  # Dropout/BN 등을 학습 모드로 전환
    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()         # 1) 이전 배치의 gradient 초기화
        logits = model(images)        # forward: 예측값 계산
        loss = criterion(logits, labels)  # loss 계산
        loss.backward()               # 2) 역전파: 각 파라미터의 gradient 계산
        optimizer.step()              # 3) gradient 이용해 파라미터 업데이트
```

## 문제 4. SGD + Momentum Optimizer 생성

### 핵심 개념: SGD with Momentum과 Weight Decay

- **정의(definition):** 기본 SGD는 `w = w - lr * g`로 파라미터를 갱신하지만, **Momentum**은 과거 gradient의 지수이동평균을 관성처럼 사용합니다: `v = μ * v + g`, `w = w - lr * v`. **Weight Decay(=L2 정규화)**는 매 step마다 `w`에 `- lr * λ * w`를 추가로 빼서 가중치가 과하게 커지는 것을 막습니다.
- **원리:** Momentum은 진동을 줄이고 평탄 구간에서 가속을 도와 수렴 속도를 높입니다. Weight decay는 과적합을 완화해 일반화 성능을 개선합니다.
- **예시:** 영상 분류에서 흔히 쓰는 설정은 `lr=0.1, momentum=0.9, weight_decay=5e-4`로 시작하여 cosine schedule로 lr을 줄이는 방식입니다.


```python
import torch
import torch.nn as nn

criterion = nn.CrossEntropyLoss()  # 분류 문제의 표준 loss (softmax + NLL 내장)

# SGD with Momentum + Weight Decay
# - lr: 학습률
# - momentum: 과거 gradient 관성 비율 (보통 0.9)
# - weight_decay: L2 정규화 계수 (가중치 크기 억제)
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.1,
    momentum=0.9,
    weight_decay=5e-4,
)
```

## 문제 5. ViT 패치 임베딩 (einops Rearrange 패턴)

### 핵심 개념: ViT의 패치 토큰화(Patchify)

- **정의(definition):** Vision Transformer(ViT)는 이미지를 **고정 크기 패치**로 나눠 각 패치를 1D 벡터로 펼친 후, 이들을 토큰 시퀀스로 취급합니다.
- **원리:** 입력 이미지 `(B, C, H, W)` 를 `P×P` 패치로 자르면 패치 개수는 `N = (H/P) × (W/P)`, 패치 벡터의 차원은 `P*P*C`가 됩니다. einops `Rearrange` 패턴 `"b c (h p1) (w p2) -> b (h w) (p1 p2 c)"`는 이를 `(B, N, P*P*C)` 모양의 토큰 시퀀스로 변환합니다. 이후 `nn.Linear(patch_dim, dim)`으로 hidden 차원에 투영합니다.
- **예시:** CIFAR-10 (32×32, 3채널)을 4×4 패치로 쪼개면 토큰 개수 N=64, 패치 차원=4·4·3=48. `Linear(48, 64)`로 투영하면 `(B, 64, 64)` 토큰 시퀀스가 됩니다.


```python
from einops.layers.torch import Rearrange
import torch.nn as nn

patch_height = 4
patch_width = 4
patch_dim = 3 * patch_height * patch_width  # 한 패치를 펼친 벡터 차원 = C * P * P
dim = 64  # Transformer hidden size

# (B, C, H, W) 이미지를 패치 토큰 시퀀스로 변환
to_patch_embedding = nn.Sequential(
    # h,w: 패치 그리드 개수 / p1,p2: 패치 내부 픽셀
    # "b c (h p1) (w p2) -> b (h w) (p1 p2 c)"
    #  - (h p1) = 원본 H,  (w p2) = 원본 W
    #  - 결과: (B, N=h*w, P*P*C)
    Rearrange("b c (h p1) (w p2) -> b (h w) (p1 p2 c)",
              p1=patch_height, p2=patch_width),
    nn.Linear(patch_dim, dim),  # 패치 벡터 -> hidden dim 으로 선형 투영
)
```

## 문제 6. Scaled Dot-Product Attention (Q·Kᵀ 유사도와 softmax)

### 핵심 개념: Scaled Dot-Product Attention

- **정의(definition):** Attention은 쿼리(Q)가 각 키(K)와 얼마나 유사한지 내적으로 계산해 가중치를 얻고, 그 가중치로 값(V)들을 가중합하는 연산입니다: `Attn(Q,K,V) = softmax(Q·Kᵀ / √d_k) · V`.
- **원리:**
    - Q·Kᵀ: 두 벡터의 내적이 클수록 유사합니다. 토큰 쌍별 유사도 행렬(shape: `(N, N)`)이 만들어집니다.
    - √d_k 로 나누는 이유: 차원이 커지면 내적 값의 분산이 커져 softmax가 한 토큰에 치우쳐 gradient가 사라집니다. 스케일링은 분포를 안정화합니다.
    - softmax(dim=-1): 각 쿼리에 대한 키 가중치의 합이 1이 되도록 확률화합니다.
- **예시:** d_k=64 라면 스케일 factor = 1/√64 = 0.125. 이 스케일이 없으면 학습 초기에 attention이 하나의 토큰만 강하게 보고 나머지는 0에 가까워지는 문제가 발생합니다.


```python
import torch
from torch import nn, einsum
from einops import rearrange

class Attention(nn.Module):
    def __init__(self, dim, heads=4, dim_head=64):
        super().__init__()
        inner_dim = dim_head * heads
        self.heads = heads
        self.scale = dim_head ** -0.5  # 1/sqrt(d_k) 스케일 factor
        self.attend = nn.Softmax(dim=-1)
        self.to_qkv = nn.Linear(dim, inner_dim * 3, bias=False)
        self.to_out = nn.Linear(inner_dim, dim)

    def forward(self, x):
        b, n, _, h = *x.shape, self.heads  # x: (B, N, D)
        qkv = self.to_qkv(x).chunk(3, dim=-1)  # Q,K,V 를 한 번에 선형변환 후 분할
        # 멀티 헤드로 쪼개기: (B, N, heads*d) -> (B, heads, N, d)
        q, k, v = map(lambda t: rearrange(t, 'b n (h d) -> b h n d', h=h), qkv)

        # Q·Kᵀ: 토큰 간 유사도. einsum 'b h i d, b h j d -> b h i j'
        # scale 로 나누어 softmax 안정화
        dots = einsum('b h i d, b h j d -> b h i j', q, k) * self.scale
        # 마지막 축(Key 방향)에 softmax -> 각 쿼리의 attention 분포(합=1)
        attn = self.attend(dots)

        # attn · V : 값 벡터의 가중합 -> 새로운 토큰 표현
        out = einsum('b h i j, b h j d -> b h i d', attn, v)
        out = rearrange(out, 'b h n d -> b n (h d)')  # head 병합
        return self.to_out(out)
```

## 문제 7. ViT CLS 토큰 결합 & 위치 임베딩 더하기

### 핵심 개념: [CLS] 토큰과 위치 임베딩(Positional Embedding)

- **정의(definition):**
    - **[CLS] 토큰**: 이미지 전체의 표현을 담기 위한 **학습 가능한 특수 토큰** 1개를 토큰 시퀀스 맨 앞에 추가합니다. Transformer 통과 후 이 토큰 벡터만 가지고 분류를 수행합니다.
    - **Positional Embedding**: Transformer의 self-attention은 순서를 모르므로, 각 토큰 위치마다 고유한 학습 벡터를 더해 공간 위치 정보를 주입합니다.
- **원리:** CLS 토큰은 self-attention 연산을 거치면서 다른 모든 패치 토큰의 정보를 흡수합니다. BERT의 [CLS]와 동일한 아이디어이며, 최종 분류 헤드는 `x[:, 0]` (CLS 토큰)만 사용합니다.
- **예시:** 패치 N=64개 → CLS 추가로 N+1=65개 토큰 → pos_embedding shape=`(1, 65, dim)`을 브로드캐스트로 더함.


```python
import torch
from torch import nn
from einops import repeat

class ViTForward(nn.Module):
    def __init__(self, num_patches, dim):
        super().__init__()
        # 학습 가능한 CLS 토큰 파라미터 (1, 1, dim)
        self.cls_token = nn.Parameter(torch.randn(1, 1, dim))
        # 학습 가능한 위치 임베딩 (1, N+1, dim) -> CLS 포함 길이
        self.pos_embedding = nn.Parameter(torch.randn(1, num_patches + 1, dim))

    def forward(self, x):
        b, n, _ = x.shape  # x: (B, N, dim) 패치 토큰 시퀀스
        # 배치 개수만큼 CLS 토큰을 복제: (1,1,d) -> (B,1,d)
        cls_tokens = repeat(self.cls_token, '1 1 d -> b 1 d', b=b)
        # CLS 토큰을 시퀀스 맨 앞에 concat: (B, N+1, dim)
        x = torch.cat((cls_tokens, x), dim=1)
        # 위치 임베딩 더해 순서 정보 주입 (브로드캐스트로 배치에 공통 적용)
        x = x + self.pos_embedding[:, :(n + 1)]
        return x
```

## 문제 8. ViT 분류 표현 추출: CLS 토큰 선택 vs mean pooling

### 핵심 개념: 시퀀스 풀링(CLS vs mean)

- **정의(definition):** Transformer 출력 `(B, N+1, dim)` 에서 분류용 **단일 벡터**를 뽑는 방법은 크게 두 가지입니다.
    - `pool == "cls"`: 맨 앞의 CLS 토큰만 사용 → `x[:, 0]`
    - `pool == "mean"`: 전체 토큰을 평균 → `x.mean(dim=1)`
- **원리:** CLS 토큰은 학습되면서 '요약자' 역할을 맡도록 최적화되고, mean pool은 모든 토큰에 동등한 가중치를 부여합니다. 데이터·하이퍼파라미터에 따라 어느 쪽이 더 좋은지 달라집니다.
- **예시:** 원 ViT 논문은 CLS 방식을, Scaling ViT·BEiT 계열에서는 mean pool 혹은 GAP(global average pooling)을 쓰기도 합니다.


```python
def select_feature(x, pool):
    # x: Transformer 출력 (B, N+1, dim)
    # pool == "cls" : 0번 위치(CLS 토큰) 벡터만 사용 -> (B, dim)
    # pool == "mean": 모든 토큰의 평균 풀링 -> (B, dim)
    x = x[:, 0] if pool == "cls" else x.mean(dim=1)
    return x
```

### 핵심 개념: 사전학습 모델을 위한 전처리 정합성

- **정의(definition):** 사전학습된 모델을 추론에 쓰려면 학습 시 사용한 **동일한 전처리(resize, tensor 변환, 정규화)**를 적용해야 합니다.
- **원리:** DETR은 ResNet backbone을 ImageNet 통계로 사전학습 후 COCO로 fine-tune 했기 때문에 `mean=(0.485, 0.456, 0.406)`, `std=(0.229, 0.224, 0.225)` 로 정규화해야 backbone 분포가 맞습니다. 공식 데모는 짧은 변을 800픽셀로 `Resize` 합니다.
- **예시:** 전처리가 다르면 같은 이미지에서도 전혀 다른 예측이 나옵니다. 예를 들어 정규화를 빼먹으면 객체 탐지 신뢰도가 크게 낮아집니다.


```python
import torchvision.transforms as T

# DETR 공식 데모 전처리 파이프라인
transform = T.Compose([
    T.Resize(800),          # 짧은 변을 800픽셀로 리사이즈
    T.ToTensor(),           # PIL 이미지 -> Tensor (C,H,W), [0,1]
    T.Normalize([0.485, 0.456, 0.406],
                [0.229, 0.224, 0.225])  # ImageNet mean/std 정규화
])
```

## 문제 10. 박스 좌표 변환: (cx,cy,w,h) → (xmin,ymin,xmax,ymax)

### 핵심 개념: 바운딩 박스 표현 방식(cxcywh ↔ xyxy)

- **정의(definition):**
    - `cxcywh`: 중심 좌표(cx,cy)와 너비(w), 높이(h). DETR·YOLO 계열 모델의 내부 표현.
    - `xyxy`: 좌상단(xmin, ymin), 우하단(xmax, ymax) 꼭짓점. 대부분의 시각화/NMS 라이브러리가 사용.
- **원리:** 중심과 크기로부터 꼭짓점은 `xmin = cx - w/2`, `ymin = cy - h/2`, `xmax = cx + w/2`, `ymax = cy + h/2`로 직관적으로 계산됩니다.
- **예시:** 중심 (100,120), w=40, h=30 박스 → `(80,105,120,135)`. DETR의 pred_boxes는 추가로 이미지 크기로 **정규화**된 값(0~1)이므로 픽셀로 바꿀 땐 (W,H,W,H)를 곱해야 합니다.


```python
import torch

def box_cxcywh_to_xyxy(x):
    # x: (N, 4), 각 row = (cx, cy, w, h)
    x_c, y_c, w, h = x.unbind(1)  # 컬럼별 분리
    # 중심±절반너비 공식으로 좌상/우하 꼭짓점 계산
    b = [(x_c - 0.5 * w),   # xmin
         (y_c - 0.5 * h),   # ymin
         (x_c + 0.5 * w),   # xmax
         (y_c + 0.5 * h)]   # ymax
    return torch.stack(b, dim=1)  # (N,4) 재결합
```

## 문제 11. DETR 예측 신뢰도 필터링

### 핵심 개념: DETR의 set prediction과 no-object 클래스

- **정의(definition):** DETR은 NMS 없이 고정 개수(default 100)의 object query가 각각 `(C+1)` 클래스 로짓과 박스를 예측합니다. **+1은 "no-object"** 클래스로, 실제 객체가 없는 query가 배정됩니다.
- **원리:** softmax를 적용한 후 마지막(no-object) 클래스를 제외한 `C`개 클래스에서 최댓값을 취하면 "실제 객체일 확률"을 얻을 수 있습니다. 이 값이 임계치(예: 0.9) 이상인 query만 남기면 자연스럽게 유의미한 예측만 필터링됩니다.
- **예시:** pred_logits shape = `(1, 100, 92)` → softmax → 마지막 축 제외 `[..., :-1]` → shape `(100, 91)` → max score가 0.9 이상인 query들만 선택.


```python
# pred_logits: (1, num_queries, num_classes+1)
# softmax 후 마지막 클래스(no-object) 제거: (num_queries, num_classes)
probas = outputs['pred_logits'].softmax(-1)[0, :, :-1]

# 각 query 의 최대 class score가 임계값(0.9)을 넘는 query만 True
# keep shape: (num_queries,) bool 마스크
keep = probas.max(-1).values > 0.9
```

## 문제 12. Forward Hook으로 중간 텐서 캡처

### 핵심 개념: PyTorch Forward Hook

- **정의(definition):** `register_forward_hook(fn)` 은 특정 `nn.Module`의 forward가 끝날 때마다 `fn(module, input, output)`을 호출하도록 등록하는 기능입니다. 모델 코드를 수정하지 않고 **중간 출력**을 엿볼 수 있습니다.
- **원리:** hook 함수가 `output`을 리스트에 append하면, forward를 한 번 돌린 뒤 해당 리스트에서 중간 feature map이나 attention weight를 얻을 수 있습니다. 분석이 끝나면 반드시 `hook.remove()`로 해제해 메모리 누수·중복 캡처를 막아야 합니다.
- **예시:** DETR에서는 backbone의 feature map, encoder 마지막 layer의 self-attention, decoder 마지막 layer의 cross-attention을 이 방식으로 뽑아 시각화합니다.


```python
# 중간 텐서 저장 리스트
conv_features, enc_attn_weights, dec_attn_weights = [], [], []

hooks = [
    # backbone 마지막에서 두 번째 모듈에 hook 등록 -> feature map 캡처
    model.backbone[-2].register_forward_hook(
        lambda self, input, output: conv_features.append(output)
    ),
    # encoder 마지막 layer 의 self-attention weight 캡처 (output[1] = attention 가중치)
    model.transformer.encoder.layers[-1].self_attn.register_forward_hook(
        lambda self, input, output: enc_attn_weights.append(output[1])
    ),
    # decoder 마지막 layer 의 cross-attention weight 캡처
    model.transformer.decoder.layers[-1].multihead_attn.register_forward_hook(
        lambda self, input, output: dec_attn_weights.append(output[1])
    ),
]

outputs = model(img)  # hook 등록 상태에서 forward 1회

# 메모리 누수와 중복 캡처 방지를 위해 모두 해제
for hook in hooks:
    hook.remove()
```

## 문제 13. U-Net DoubleConv 블록 구현

### 핵심 개념: U-Net의 DoubleConv(Conv-ReLU-Conv-ReLU)

- **정의(definition):** U-Net의 기본 블록은 `Conv3x3 → 활성화 → Conv3x3 → 활성화`의 2단 합성곱 구조입니다. 한 블록 안에서 채널 수를 원하는 `out_channels`로 바꾸고 특징을 정제합니다.
- **원리:** 3x3 Conv 2개를 쌓으면 receptive field는 5x5와 같아집니다. 더 적은 파라미터로 더 넓은 영역을 보면서 비선형성을 두 번 통과해 표현력을 높이는 전형적인 VGG 스타일 패턴입니다. padding=1로 공간 해상도(H,W)는 유지됩니다.
- **예시:** `(B, 64, H, W) → DoubleConv(64,128) → (B, 128, H, W)`. 이후 MaxPool로 `H,W`를 절반으로 줄이는 Down 블록과 짝지어집니다.


```python
import torch.nn as nn

class DoubleConv(nn.Module):
    """U-Net 기본 블록: (Conv3x3 -> ReLU) x 2"""
    def __init__(self, in_channels, out_channels, mid_channels=None):
        super().__init__()
        if mid_channels is None:
            mid_channels = out_channels
        self.net = nn.Sequential(
            # 1) C_in -> C_mid (padding=1 로 H,W 유지)
            nn.Conv2d(in_channels, mid_channels, kernel_size=3, padding=1, bias=False),
            nn.ReLU(inplace=True),
            # 2) C_mid -> C_out
            nn.Conv2d(mid_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.ReLU(inplace=True),
        )

    def forward(self, x):
        return self.net(x)
```

## 문제 14. U-Net Skip Connection 결합

### 핵심 개념: Skip Connection과 공간 정보 복원

- **정의(definition):** U-Net의 Skip Connection은 Encoder 단계에서 저장한 feature map `x2`를, 같은 해상도로 업샘플된 Decoder feature `x1`에 **채널 방향으로 concat** 하는 연결입니다.
- **원리:** Encoder는 반복된 pooling으로 공간 해상도를 잃기 때문에 세밀한 경계 정보가 사라집니다. Skip은 사라지기 전의 고해상도 특징을 그대로 가져와 Decoder가 정확한 위치 정보를 복원하도록 돕습니다. 두 텐서를 `dim=1` (채널 축)으로 concat하므로 출력 채널은 `C_enc + C_dec`가 됩니다.
- **예시:** `x1=(B, 256, 32, 32)`, `x2=(B, 256, 32, 32)` → `torch.cat([x2, x1], dim=1)` → `(B, 512, 32, 32)`. 이후 DoubleConv가 다시 채널을 줄입니다.


```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class Up(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        # ConvTranspose2d 로 공간 해상도 2배 업샘플 + 채널 절반으로
        self.up = nn.ConvTranspose2d(in_channels, in_channels // 2,
                                     kernel_size=2, stride=2)
        self.conv = DoubleConv(in_channels, out_channels)

    def forward(self, x1, x2):
        # x1: 디코더 쪽 (저해상도) / x2: 인코더 skip (고해상도)
        x1 = self.up(x1)  # 해상도를 skip과 맞춤

        # 홀수 해상도 입력 등으로 인한 1~2픽셀 편차 보정
        diff_y = x2.size(2) - x1.size(2)
        diff_x = x2.size(3) - x1.size(3)
        x1 = F.pad(x1, [diff_x // 2, diff_x - diff_x // 2,
                        diff_y // 2, diff_y - diff_y // 2])

        # 채널 축(dim=1)으로 skip concat : (B, C_skip + C_up, H, W)
        x = torch.cat([x2, x1], dim=1)
        return self.conv(x)  # DoubleConv 로 채널 정제 및 축소
```

## 문제 15. Binary Segmentation 예측 마스크 생성

### 핵심 개념: BCEWithLogitsLoss와 추론 시 sigmoid + threshold

- **정의(definition):** Binary Segmentation에서 출력은 채널 1개짜리 **logits**(무제한 실수)입니다. 학습엔 `nn.BCEWithLogitsLoss`(sigmoid+BCE를 수치 안정적으로 결합)를 쓰고, 추론에서는 `sigmoid`로 확률화한 뒤 임계값 0.5로 이진화해 픽셀 마스크를 만듭니다.
- **원리:** 학습 시 `BCEWithLogitsLoss`는 내부적으로 log-sum-exp 트릭을 사용해 sigmoid를 따로 적용할 필요가 없습니다. 추론 시에만 sigmoid를 적용하고 임계값으로 binary 마스크를 얻습니다.
- **예시:** logits = 2.0 → sigmoid ≈ 0.88 → (>0.5) → 1. logits = -1.5 → sigmoid ≈ 0.18 → 0.


```python
import torch

@torch.no_grad()
def predict_mask(model, imgs):
    model.eval()
    logits = model(imgs)          # (B, 1, H, W) logits
    probs = torch.sigmoid(logits) # 로짓을 [0,1] 확률로 변환
    preds = (probs > 0.5).float() # 0.5 임계값으로 이진 마스크 생성
    return preds
```

## 문제 16. DDPM Sinusoidal Timestep Embedding

### 핵심 개념: 확산 모델의 타임스텝 임베딩(Sinusoidal)

- **정의(definition):** 확산 모델(DDPM)은 `t=0..T` 단계 중 '현재 몇 번째 스텝인지'를 네트워크에 알려야 합니다. Transformer의 positional encoding과 같이 `sin/cos` 파형 여러 주파수를 이어붙인 벡터로 정수 t를 고차원 임베딩으로 바꿉니다.
- **원리:** 주파수가 다른 여러 사인/코사인의 값은 t마다 고유한 패턴을 만들어 네트워크가 t를 구별할 수 있게 합니다. half_dim 개의 서로 다른 주파수를 만든 뒤 `sin`, `cos`을 concat하여 `dim` 차원을 채웁니다. 주파수는 `exp(-log(10000) * k / (half_dim-1))` 형태.
- **예시:** dim=4, t=10 → half_dim=2, 주파수=[1.0, 0.0001], args=[10, 0.001] → `[sin(10), sin(0.001), cos(10), cos(0.001)]`.


```python
import math
import torch
import torch.nn as nn

class SinusoidalPositionEmbeddings(nn.Module):
    """Transformer-style sinusoidal embedding을 timestep 인코딩으로 활용"""
    def __init__(self, dim):
        super().__init__()
        self.dim = dim  # 최종 임베딩 차원

    def forward(self, time):
        device = time.device
        half_dim = self.dim // 2
        # 로그 스케일로 감소하는 주파수 시퀀스 생성 (1 -> 0.0001)
        freq = math.log(10000) / (half_dim - 1)
        freq = torch.exp(torch.arange(half_dim, device=device) * -freq)
        # (batch, half_dim) : 각 t에 대해 주파수별 각도
        args = time[:, None].float() * freq[None, :]
        # sin/cos 값을 concat -> (batch, dim) 임베딩 완성
        emb = torch.cat([args.sin(), args.cos()], dim=-1)
        return emb
```

## 문제 17. DDPM Timestep 주입(FiLM / AdaGN scale-shift)

### 핵심 개념: FiLM (Feature-wise Linear Modulation)

- **정의(definition):** 조건 변수(예: timestep t)를 네트워크에 주입할 때, 단순히 더하는 대신 채널별로 **스케일(scale)과 시프트(shift)**를 만들어 feature map을 `x = x * (scale + 1) + shift`로 변조하는 기법입니다.
- **원리:** 더하기만 하면 조건이 feature 크기와 섞이기 어렵지만, 곱셈은 "조건에 따라 특정 채널을 강조/억제"하는 게이팅 효과를 줘서 훨씬 강력합니다. `scale+1`로 두는 이유는 초기값이 0에 가까워도 `x * 1 = x`가 되어 학습 초기에 안정적으로 시작하기 때문입니다.
- **예시:** U-Net의 각 ResNet block에서 timestep MLP가 `(B, 2C)`를 만들어 `chunk(2)` 로 `(scale, shift)`를 얻은 뒤 GroupNorm 뒤의 feature map에 적용합니다.


```python
import torch.nn as nn

class Block(nn.Module):
    """Conv -> GroupNorm -> (FiLM scale/shift) -> SiLU"""
    def __init__(self, dim, dim_out, groups=8):
        super().__init__()
        self.proj = nn.Conv2d(dim, dim_out, 3, padding=1)
        self.norm = nn.GroupNorm(groups, dim_out)
        self.act = nn.SiLU()

    def forward(self, x, scale_shift=None):
        x = self.proj(x)
        x = self.norm(x)
        if scale_shift is not None:
            scale, shift = scale_shift  # 각각 (B, C, 1, 1)
            # FiLM: 조건(timestep) 기반으로 채널별 affine 변환
            # (scale + 1) 로 두어 초기 weight≈0 일 때 x * 1 = x 로 시작 (안정)
            x = x * (scale + 1) + shift
        x = self.act(x)
        return x
```

## 문제 18. DDPM U-Net의 Self-Attention(einsum)

### 핵심 개념: Conv feature map에 대한 Self-Attention

- **정의(definition):** Conv feature map `(B, C, H, W)`에 대해 공간 축(H·W)을 토큰으로 보고 Self-Attention을 적용하면 모든 픽셀 쌍의 장거리 관계를 학습할 수 있습니다.
- **원리:** Q,K,V를 1x1 Conv로 만든 뒤 `(B, heads, d, HW)` 형태로 재배열합니다. `einsum('b h d i, b h d j -> b h i j', q, k)`는 각 쿼리 위치 i와 키 위치 j 사이의 내적 행렬(`(HW, HW)`)을 계산합니다. softmax 후 `einsum('b h i j, b h d j -> b h i d', attn, v)`로 값의 가중합을 구하고 다시 `(B, C, H, W)`로 복원합니다.
- **예시:** 16x16 feature map에서 HW=256, attention matrix 크기는 `256x256`. 해상도가 커질수록 `O(N²)`로 비용이 급증해 DDPM에서는 저해상도 구간에만 이런 "full attention"을 씁니다.


```python
import torch
import torch.nn as nn
from einops import rearrange

class DDPMAttention(nn.Module):
    """Conv feature map(B,C,H,W) 위에서의 Multi-Head Self-Attention"""
    def __init__(self, dim, heads=4, dim_head=32):
        super().__init__()
        self.scale = dim_head ** -0.5  # 1/sqrt(d_k)
        self.heads = heads
        hidden_dim = dim_head * heads
        self.to_qkv = nn.Conv2d(dim, hidden_dim * 3, 1, bias=False)  # 1x1 conv 로 Q,K,V 동시 생성
        self.to_out = nn.Conv2d(hidden_dim, dim, 1)

    def forward(self, x):
        b, c, h, w = x.shape
        # (B, 3*hidden, H, W) -> 3 개의 (B, hidden, H, W)
        qkv = self.to_qkv(x).chunk(3, dim=1)
        # head 로 쪼개고 공간(H,W)을 시퀀스로 flatten: (B, heads, d, HW)
        q, k, v = map(lambda t: rearrange(t, 'b (h c) x y -> b h c (x y)', h=self.heads), qkv)

        q = q * self.scale
        # Q·Kᵀ : 쿼리 위치 i 와 키 위치 j 사이 내적 -> (B, heads, HW, HW)
        sim = torch.einsum('b h d i, b h d j -> b h i j', q, k)

        # softmax 수치 안정화: 최대값을 빼서 지수 폭주 방지 (detach로 그래프 분리)
        sim = sim - sim.amax(dim=-1, keepdim=True).detach()
        attn = sim.softmax(dim=-1)  # key 축에 softmax

        # attn·V: 값의 가중합 -> (B, heads, HW, d)
        out = torch.einsum('b h i j, b h d j -> b h i d', attn, v)
        # 다시 이미지 공간으로 복원: (B, hidden_dim, H, W)
        out = rearrange(out, 'b h (x y) d -> b (h d) x y', x=h, y=w)
        return self.to_out(out)
```

## 문제 19. Stable Diffusion 파이프라인으로 이미지 생성

### 핵심 개념: Classifier-Free Guidance Scale

- **정의(definition):** Stable Diffusion은 prompt를 조건으로 디노이징을 수행합니다. `guidance_scale`은 "프롬프트를 얼마나 강하게 반영할지" 조절하는 하이퍼파라미터로, **Classifier-Free Guidance(CFG)**의 가중치 w입니다. 수식: `noise = uncond + w*(cond - uncond)`.
- **원리:** `w=1`이면 조건만 그대로 사용, `w>1`이면 조건 방향을 더 과장합니다. 너무 작으면(≈1) 프롬프트 반영이 약하고, 너무 크면(>20) 채도·대비가 과해져 품질이 깨집니다. 실무 권장값은 `7~8.5` 정도입니다.
- **예시:** `pipe("A cat", guidance_scale=1.5)` 느슨한 반영, `guidance_scale=30` 프롬프트에 과적합된 거친 이미지.


```python
from diffusers import StableDiffusionPipeline
import torch

# Stable Diffusion v1-4 파이프라인 로드 (float16 로 VRAM 절약)
pipe = StableDiffusionPipeline.from_pretrained(
    "CompVis/stable-diffusion-v1-4",
    torch_dtype=torch.float16
).to("cuda")

prompt = "A futuristic cityscape at sunset with flying cars"
# pipe(prompt, guidance_scale=...).images[0]:
# - StableDiffusionPipelineOutput.images 는 PIL.Image 리스트
# - guidance_scale=7.5 는 프롬프트 반영 강도의 권장값
image = pipe(prompt, guidance_scale=7.5).images[0]
```

## 문제 20. 재현 가능한 SD 생성: generator seed + negative_prompt

### 핵심 개념: 재현성(Seed)과 Negative Prompt

- **정의(definition):**
    - **Generator seed**: diffusion 과정은 초기 noise의 랜덤 샘플링에 의존합니다. `torch.Generator("cuda").manual_seed(s)`로 생성자 RNG를 고정하면 **같은 조건에서 완전히 동일한 이미지**가 만들어져 비교 실험이 가능해집니다.
    - **Negative prompt**: 생성 결과에서 **빠지길 원하는 개념**을 별도로 지정합니다. CFG 수식에서 "무조건(unconditional)" 역할이 기본 빈 문자열 대신 negative prompt 임베딩으로 대체되는 효과가 있어, 해당 개념이 적극적으로 억제됩니다.
- **원리:** Seed를 고정하면 초기 latent noise가 같아지므로 guidance_scale·prompt 등 다른 변수의 영향만 비교할 수 있습니다. Negative prompt는 "여자를 그리지 마", "빨간색을 쓰지 마"처럼 사용합니다.
- **예시:** `pipe("forest", negative_prompt="person, red", generator=torch.manual_seed(42))` 숲 이미지에 사람·빨간 객체를 억제.


```python
import torch

prompt = "A beautiful watercolor painting of a green forest"
negative = "woman, red flowers, red color"

# 재현성을 위해 CUDA 생성자 RNG 시드를 고정
# 동일한 seed + 동일한 prompt/설정 -> 항상 같은 이미지
generator = torch.Generator("cuda").manual_seed(2024)

image = pipe(
    prompt,
    negative_prompt=negative,       # 프롬프트에서 배제할 개념 지정
    generator=generator,            # 고정 시드 생성자 전달
    num_inference_steps=30,         # 디노이징 스텝 수(품질↔속도)
    guidance_scale=7.5,             # CFG 가중치 (권장값)
).images[0]
```
